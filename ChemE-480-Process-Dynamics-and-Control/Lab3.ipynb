## EXERCISE 1: FITTING FIRST-ORDER PLUS TIME DELAY SYSTEM

The Q1/T1 system can be approximated fairly well by a first-order model. 

A first-order dynamic model of the Temperature Control Lab has the form:

$$\tau_p \frac{dT'}{dt} = -T' + K_pQ'(t - \theta_p) $$

where $T' = T - T_{ss}$ and $Q'=Q - Q_{ss}$ are deviation variables with steady-state initial conditions. Deviation variables are how we re-adjust such that we are able often set our initial conditions equal to 0 since at time $t=0$, if the system starts at steady-state, $T(0)' = T_{ss} - T_{ss} = 0$. Here $\tau_p$ is a time constant that dictates the speed of the response to change, $K_p$ is the gain (or the amount of output change for a given input change), and $(t-\theta_p)$ indicates a time delay of length $\theta_p$. ### SAVE DATA FOR LATER USE

Create a function to save the data from the Arduino to a file.

```python
def save_txt(t, u1, u2, y1, y2):
    '''
    The goal of this function is to save the data into a txt file.
    
    INPUTS
    
        t (array) - time array
        u1 (array) - input 1 array
        u2 (array) - input 2 array
        y1 (array) - output 1 array
        y2 (array) - output 2 array
    
    OUTPUTS
    
        none
    '''
    
    # Create a vertical stack
    data = np.vstack((t, u1, u2, y1, y2))
    
    # Transpose the data
    data = data.T
    
    #
    top = ('Time,Q1,Q2,T1,T2')
    np.savetxt('foptd.txt', data, delimiter=',',header=top, comments='')
```
