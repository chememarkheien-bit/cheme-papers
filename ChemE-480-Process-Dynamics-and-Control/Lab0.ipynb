## PROGRAM TESTING

You can test your installation with the code below:

```python
import tclab
with tclab.TCLab() as lab:
    print(lab.T1)
```

If the above step is run when you don't have an Arduino present, but you do have `tclab` installed correctly, you should see the following error message. This is what you should see today.

    RuntimeError: No Arduino device found.


If the above step is run when you actually have an Arduino present, you should see the following message:


    TCLab version 1.0.0
    Arduino Leonardo connected on port /dev/cu.usbmodem14201 at 115200 baud.
    TCLab Firmware 2.0.1 Arduino Leonardo/Micro.
    22.445
    TCLab disconnected successfully.
