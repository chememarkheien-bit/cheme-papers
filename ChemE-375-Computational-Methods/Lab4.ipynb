```python
import numpy as np

# Define the matrix A using Numpy arrays
A = np.array([[1, 0, -1, 0],
              [4, 0, 0, -2],
              [0, 2, -2, -1],
              [1, 0, 0, 0]])

# Define the matrix B
b = np.array([0, 0, 0, 1])

# Calculate the inverse of A and multiply it by b
x = np.linalg.inv(A).dot(b)

# Print the solution
print(f"The solution is {x}")
```
